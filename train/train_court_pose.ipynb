## 학습
화면 전체의 코트 기하가 중요하므로 mosaic과 회전 증강은 끕니다.

**좌우 반전 증강은 끕니다 (fliplr=0.0).** 모델에는 한 가지 배치(보이는 사이드라인이 화면 오른쪽)만 가르칩니다. 좌우가 반대인 촬영(1103)은 make_dataset이 프레임과 라벨을 미리 뒤집어 이 배치로 맞추고, 실제 사용 때는 keypoint_init이 원본과 반전 프레임을 모두 풀어 흰 선에 더 잘 맞는 쪽을 씁니다. 두 배치를 한 모델에 섞으면 왼쪽/오른쪽 짝을 섞어 냅니다 (2026-11-11 fliplr=0.5 학습: 검증 지표는 높았지만 1008 화면에서 far_corner_right 754 px, near_10m 1064 px 오차).

1103은 카메라가 더 높고 멀어서 크기·원근 증강을 키웠습니다 (scale 0.3, perspective 0.0003).

```python
from ultralytics import YOLO

model = YOLO('yolo11n-pose.pt')
model.train(
    data='/content/court_pose/data.yaml',
    imgsz=960, epochs=100, batch=16, patience=25,
    mosaic=0.0, degrees=0.0, shear=0.0, perspective=0.0003,
    translate=0.1, scale=0.3, fliplr=0.0,
    hsv_h=0.015, hsv_s=0.5, hsv_v=0.4,
    project=DRIVE + '/runs', name='court_pose', exist_ok=True, save_period=10,
)
```
