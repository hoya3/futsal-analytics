## 학습
화면 전체의 코트 기하가 중요하므로 mosaic과 회전 증강은 끕니다. 좌우 반전도 끕니다. 두 카메라 모두 대각선 사이드라인이 화면 오른쪽에 보여 반전 영상은 실제로 나오지 않고, 지점 이름이 카메라 기준 왼쪽/오른쪽이라 반전을 켜면 먼 쪽 오른쪽 지점(코너, 호, 골포스트)을 잘못 배운다 (fliplr=0.5로 학습한 모델에서 화면 오차 560~690 px).


```python
from ultralytics import YOLO

model = YOLO('yolo11n-pose.pt')
model.train(
    data='/content/court_pose/data.yaml',
    imgsz=960, epochs=100, batch=16, patience=25,
    mosaic=0.0, degrees=0.0, shear=0.0, perspective=0.0,
    translate=0.1, scale=0.1, fliplr=0.0,
    hsv_h=0.015, hsv_s=0.5, hsv_v=0.4,
    project=DRIVE + '/runs', name='court_pose', exist_ok=True, save_period=10,
)
```
